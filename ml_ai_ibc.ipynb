```python
correct_guess = 0
total_guesses = 5
for guesses in range(1,total_guesses + 1):
    number_to_guess = int(input('Enter Your Number: '))
    if number_to_guess == 7:
        print(f'You have guessed right.')
        correct_guess += 1
    else:
        print(f'You have guessed wrong.')
    
    guess_left = total_guesses - guesses
    print(f'You have {guess_left} guesses left')
    guess_left -= 1
print(f"You got {correct_guess} correct guesses")

    
```

```python
#ASSIGNMENT 1
city_name = str(input('Enter your city name: '.title()))
week_number = int(input('Enter the week number: '.title()))
print(f'{city_name}, Week number {week_number}')
temp = []
for degrees in range(1,8):
    degrees = float(input(f'Enter the temperature for day {degrees}: '.title()))
    temp.append(degrees)    

    if degrees >= 30:
        print('Hot')
    elif 20 < degrees < 30:
        print('Warm')
    else:
        print('Cool')
print(max(temp), min(temp))

add_temp = sum(temp)
average_temp = add_temp/len(temp)
print(average_temp)
print
if average_temp >= 25:
    print('This is a warm week.')
else:
    print('This is a mild week.')
```
