```python
# Just an example for data science course
def rank(data):
    sorted_data = sorted(data)
    ranks=[]
    for value in data:
        first=sorted_data.index(value)+1
        count=sorted_data.count(value)
        ranks.append(first+(count-1)/2)
    return ranks
def spearman(x,y):
    n=len(x)
    rx,ry=rank(x),rank(y)
    d_squared=sum((a-b)**2 for a,b in zip(rx,ry))
    return 1-(6*d_squared)/(n*(n**2-1))
x=[86,97,99,100,101,103,106,110,112,113,98]
y=[2,6,7,12,17,20,27,28,29,50,40]
print(f"Spearman rho = {spearman(x,y):.4f}")
```

```python
import time

song_title = "If You Like"
artist = "BNXN"

# Original lyrics (replace with your own)
lyrics = """
I will really like to be your baby
But that's if, if you like
If you have somebody else ke sọ fun me
Break my heart, if you like
Oh-oh If you like, I go like
Me and you go somewhere nice on your time
Aeroplane, straight to Mykonos
But it all depends on if you really like
"""

LINE_DELAY = 1.5   # seconds to pause after each line
CHAR_DELAY = 0.05  # seconds between each character


def type_line(line):
    for char in line:
        print(char, end="", flush=True)
        time.sleep(CHAR_DELAY)
    print()


def karaoke(title, singer, text):
    print(f"{title} - {singer}".center(50, "="))
    print()
    time.sleep(2)

    for line in text.strip().split("\n"):
        type_line(line.strip())
        time.sleep(LINE_DELAY)

    print()
    print("THE END".center(50, "="))


karaoke(song_title, artist, lyrics)
```
